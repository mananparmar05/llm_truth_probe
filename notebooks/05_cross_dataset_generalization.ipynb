# 05 — Cross-Dataset Generalisation
Train on TruthfulQA, evaluate on HaluEval (and vice versa).

```python
import sys; sys.path.insert(0, '..')
import numpy as np
from extraction.storage import HiddenStateStore
from analysis.cross_dataset import CrossDatasetStudy
import pathlib

store = HiddenStateStore("../data/hidden_states", format="hdf5")
datasets = {}
for name in ["truthfulqa", "halueval"]:
    try:
        hs, labels = store.load(name)
        datasets[name] = (hs, labels)
    except Exception as e:
        print(f"Could not load {name}: {e}")

if len(datasets) == 2:
    study = CrossDatasetStudy(datasets)
    df = study.run_all()
    print(df.pivot_table(index="train_set", columns="test_set", values="auroc", aggfunc="mean"))
    CrossDatasetStudy.plot(df, save_path=pathlib.Path("../results/plots/cross_dataset_auroc.png"))
```
